## data containers

Organize the data structure into four different families: 
- Ordered data structure:  list and tuple
- Unordered data structure:  set and dictionary 
- Mutable:  list, set and dictionary 
- Immutable:  tuple 


| Type | Example | Description |
|------|---------|---------|
| list | `[1, 2, 3]` | Ordered collection|
| tuple | `(1, 2, 3)` | Immutable ordered collection|
| dict | `{'a':1, 'b':2, 'c':3}` | Unordered (key:value) pair mapping|
| set | `{1, 2, 3}` | Unordered collection of unique values |

## Sets

The set contains unordered collections of unique items.

- use the curly brackets `{}`
- **Unordered**
- **Set elements are unique**. Duplicate elements are not allowed.
- **Mutable**: a set itself may be modified, but the elements contained in the set must be of an immutable type.
- **No Indexing**: Set items cannot be referred to by index or key.


```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}

myset = {'a', 'a', 'a', 'a', 'b', 'b', 'c'}
print(myset)
myset = {'a', 'all', 1, 2, (1, 2, 3)}
print(myset)
myset = {'a', 'all', 1, 2, [1, 2, 3]}
print(myset)
```

Sets relate very closely to the sets we know from mathematics, the familiar operations like the union, intersection, difference, symmetric difference exist.


Python's sets have all of these operations built-in, via methods or operators.

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
# union: items appearing in either
res_1o =  primes | odds      # with an operator
res_1m = primes.union(odds) # equivalently with a method
print(res_1o)
print(res_1m)
```

```python
# intersection: items appearing in both
res_2o = primes & odds             # with an operator
res_2m = primes.intersection(odds) # equivalently with a method
print(res_2o)
print(res_2m)
```

```python
# difference: items in primes but not in odds
res_3o = primes - odds           # with an operator
res_3m = primes.difference(odds) # equivalently with a method
print(res_3o)
print(res_3m)
```

```python
# symmetric difference: items appearing in only one set
res_4o = primes ^ odds                     # with an operator
res_4m = primes.symmetric_difference(odds) # equivalently with a method
print(res_4o)
print(res_4m)
```

Building an set

- an empty set with `set()`
- convert a list, tuple or string with `set()` will return a set with the elements

```python
empty_set = set()
print(empty_set)

str_1 = ' Hello World, this is a string'
str_set = set(str_1)
print(str_1)
print(str_set)
```

More set methods and operations are available.

- add: add an item to the set
- discard: remove an item
- pop: removes a random item from the set and returns its value

```python
print(primes)
primes.add(13)
print(primes)

guess_what = primes.pop()
print(guess_what)

primes.discard(1)
print(primes)

primes.remove(1)
print(primes)
```
