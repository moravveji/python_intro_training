## Basics

This document is a Jupyter notebook, it is built out of cells: text and code. The text cells can be formatted using the *markdown* syntax. The code cells contain code statements, usually Python code. 
It is an interactive environment that enbles you to execute the code cells.

## cell mode

Cells can be in 2 modes: **command** mode and **edit** mode. 

### Edit mode 

- Is indicated by a *green cell border* and a prompt showing in the editor area
- When a cell is in edit mode, you can type into the cell, like a normal text editor.
- Enter edit mode by pressing `Enter` or using the mouse to click on a cell’s editor area.

### Command mode

- Is indicated by a *grey cell border with a blue left margin*
- When you are in command mode, you are able to manipulate the notebook as a whole, but not type into the individual cell. 
- Enter command mode by pressing `Esc` or click the mouse outside the cell’s editor area.

## Execute a cell

Running a cell works for both text and code cells.

- `Shift + Enter` runs the current cell, move to cell below, add a new cell when no cell is below
- `Ctrl + Enter` runs the selected cell 
- `Alt + Enter` runs the current cell, insert a cell below

## Useful to know

- All cells modify the same global state, so variables that you define by executing a cell can be used in other cells
- `[prompt number] in front of a code cell`, indicates the execution number. The kernel should have a single, monotonically increasing counter of all execution requests
- `[*] instead of a prompt number`, indicates that the kernel is busy. Go to the menu Kernel and click Interrupt if you suspect the program . If this does not work click Restart.
- tab completion works!

### Magic commands

A magic function begins with either a `%` or `%%` sign, and let you perform various useful tasks. Those with a `%` sign work within the environment, and those with a `%%` sign work at the cell level.
Information of a specific magic function is obtained by putting a `?` behind the command e.g. %magicfunction?

- `%lsmagic`Display the list of magic functions
- `%time` Times a single statement.
- `%reset` Delete all variables and names defined in the current namespace.
- `%run` Run a python script inside a notebook. %run script.py
- `%who, %who_ls, %whos` Display variables defined in the interactive namespace, with varying levels of verbosity
- `%pwd`Display the current working directory
- `!`Run a OS command from within Jupyter

```python
%lsmagic
```

%%time is a cell magic command that measures the execution time of the cell. 

The cell contains a Python command that makes the program sleep for two seconds. 

The %%time command will measure and print the time it took to execute the cell. 

Note that cell magics only work in IPython and Jupyter Notebooks. They are not part of the standard Python language.

```python
%%time
import time
time.sleep(2)  # sleep for two seconds
```
