fill-in-the-braces

Syntax: `S.format(item0, item1, item2, ..., itemk)`

- string type has `format()` method which takes arguments and returns another string which is the original with the method’s arguments inserted into it in certain places. 
- `format()` replaces each pair of curly brackets with the corresponding argument. 


````python
s1 = 'The {} in the {}'.format('car', 'room') 
print(s1)
s2 = '{2} color {1}, {0} {3}'.format('dark', 'brown', 'yellow', 'red') 
print(s2)
s3 = '{c} color {b}, {a} {d}'.format(a='dark', b='brown', c='yellow', d='red')                         
print(s3)
````

| g | Same as "e" if exponent is greater than -4 or less than precision, "f" otherwise.|
| G | Same as "G" if exponent is greater than -4 or less than precision, "F" otherwise.|

| s | String (converts any python object using str()).|

````python
# formatted output
count = 5 
amount = 45.56 
print('count is {0} and amount is {1:9.5f}'.format(count, amount))
print('count is {0:5d} and amount is {1:9.5f}'.format(count, amount))
print('count is {:d} and amount is {:f}'.format(count, amount))
print('count is {:05d} and amount is {:09.5f}'.format(count, amount))
print('count is {:+5d} and amount is {:09.5f}'.format(count, amount))

person = 'Louis'
greeting = 'Hello, {}!'.format(person)
print(greeting)
greeting = 'Hello, {0}, Good morning, {0}!'.format(person)
print(greeting)
````

````python
# f-string
count = 5 
amount = 45.56 
print(f'count is {count} and amount is {amount:9.5f}')
print(f'count is {count:5d} and amount is {amount:9.5f}')
print(f'count is {count:d} and amount is {amount:f}')
print(f'count is {count:05d} and amount is {amount:09.5f}')
print(f'count is {count:+5d} and amount is {amount:09.5f}')

person = 'Louis'
greeting = f'Hello, {person}!'
print(greeting)
greeting = f'Hello, {person}, Good morning, {person}!'
print(greeting)
````

## File IO

### Motivation

Storing data is important when working over multiple sessions and sharing the results with collaborators. When Python closes, all the variables in the memory are lost, data must be stored in some way. 

### Steps involved
Working with files, involves the steps:
* Opening the file and associating it with a file handle. (`open`)
* Using the file handle: read from or write to a file
* Closing the file handle to commit to a disk file. (Do not forget!) (`close`)

open a file:
```python
f = open("name.txt", mode="w", encoding="utf8")
```
The parameters are:
* name of the file, it can also contain an absolute or a relative path
* `mode` argument defines how to access the file. 

Character | Meaning
:--------:|--------
'r'       | open for reading (default)
'w'       | open for writing, truncating the file first
'x'       | open for exclusive creation, failing if the file already exists
'a'       | open for writing, appending to the end of the file if it exists

* `encoding` parameter defines the character encoding. It is not mandatory.


### Writing to a file
- `f.write(string)`
   - Copies the value of string to the file
   - Returns the number of characters written

- `f.writelines(list)`
   - Copies the value of each element of sequence to the file
   - Does not add a newline after each string - must be explicitly included in the string

````python
username = input('Enter your name: ')
f = open('name_01.txt', mode='w', encoding='utf8')
numchar = f.write(username)
print(numchar)
f.close()
````

````python
# no automatic generation of 'new lines'
name1 = 'KU Leuven'
name2 = 'campus Arenberg'
listnames = [name1, name2]
f = open('name_02.txt', mode='w', encoding='utf8')
f.write(name1)
f.write(name2)
f.write('\n')
f.writelines(listnames)
f.write('\n')
# or add a new line for each item
f.writelines(line + '\n' for line in listnames)
f.close()
````

### Read from a file

* `f.read()` reads the whole file from the current position to the end of the file as a string
* `f.read(b)` reads `b` bytes of the file, as data is read, a pointer is set of where the next data will be read
* `f.readline()` reads the data from the current position in the file to the end of that line. 
* `f.readlines()` reads every line of the file and stores them in a list.

To read the whole text file line by line, you can use `for` loop with the file object:
```python
for line in f:
    print(line)
```

Caution: `read`, `readline` and `readlines` do not check memory to perform the task.

````python
txtfile = open("file_squares.txt","r")
txt = txtfile.read()
print(txt)
print('--')
# what will be printed?
txtbis = txtfile.read()
print(txtbis)
txtfile.close()
````

seek() method

`fileObject.seek(offset[, whence])`

Parameters

- offset: position of the read/write pointer within the file.
- whence: optional parameter, defaults to 0 which means absolute file positioning, 1 which means seek relative to the current position and 2 means seek relative to the file's end.

````python
txtfile = open("file_squares.txt","r")
# set the file handle to some spot in the file
txtfile.seek(50)
txt = txtfile.read()
print(txt)
txtfile.close()
````

The `with` statement automatically takes care of closing the file once it leaves the with block, even in cases of error.
no need to call `file.close()` when using with statement. 

`with` statement itself ensures proper acquisition and release of resources. 


Nice example from bernd Klein: Numerisches Python

- open a file as input
- read the data, split the lines, manipulate the data
- write the results to another file

````python
# example from bernd Klein: Numerisches Python
with open('./Data/bKlein_bundeslaender.txt') as fh_in, \
     open('./Data/bKlein_bundeslaender_bis.txt', 'w') as fh_out:
    fh_in.readline() # read header input file
    fh_out.write('Land Surface Population PopulationDensity \n')
    for line in fh_in:
       land, surf, man, women = line.split()  # get a line and split into text components
       surf = float(surf)
       population = int(man) + int(women)
       pDensity = round(population * 1000/ surf, 2)
       fh_out.write(land + " " + str(surf) + " " + str(population) + " " + \
                    str(pDensity) + "\n")
                 
````

### IO and data formats

use the __batteries included__

Python provides the csv library. While you can also just simply use `split` function, to separate lines and data within each line, the CSV module can also be used to make things easy.
Each row read from the csv file is returned as a list of strings.

https://docs.python.org/3/library/csv.html

The csv module's `reader` and `writer` objects read and write sequences. Read and write data in dictionary form using the `DictReader` and `DictWriter` classes.

- `reader`	returns a reader object which iterates over lines of a CSV file
- `writer`	returns a writer object which writes data into CSV file
- `DictReader` class operates like a regular reader but maps the information read into a dictionary. The keys for the dictionary can be inferred from the first row of the CSV file


````python
import csv

with open('MOCK_DATA.csv') as csvfile:
    readCSV = csv.reader(csvfile, delimiter=',')
    for row in readCSV:
        print(row)
        print(row[0])
````

````python
import csv

with open('MOCK_DATA.csv') as csvfile:
    readCSV = csv.reader(csvfile, delimiter=',')
    dates = []
    ipads = []
    for row in readCSV:
        date = row[6]
        ipad = row[5]
        dates.append(date)
        ipads.append(ipad)

print(dates)
print(ipads)
````

````python
import csv
file_name = 'MOCK_DATA.csv'
f_csv = open(file_name)
csv_data = csv.DictReader(f_csv)
print("Type of object csv_data: {}\n".format(type(csv_data)))
f_csv.close()
````

````python
import csv
file_name = 'MOCK_DATA.csv'
f_csv = open(file_name)
csv_data = csv.DictReader(f_csv)

line_count = 1
data_dict = dict()
for row in csv_data:

#    print('Row {:d}: {} (length: {})'.format(line_count, dict(row), len(row)))
    data_dict[line_count] = row
    line_count += 1
     
f_csv.close()

# use dictionary
print(data_dict[54])
print(data_dict[154]['last_name'])
print(data_dict[254]['ip_address'])
print(data_dict[354]['unit cost'])
````

copy data from a file to another

````python
import csv
with open('MOCK_DATA_out.csv' ,'w') as outFile:
    fileWriter = csv.writer(outFile)
    with open('MOCK_DATA.csv','r') as inFile:
        fileReader = csv.reader(inFile)
        for row in fileReader:
            fileWriter.writerow(row)
````
