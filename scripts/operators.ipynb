# Operators
## Arithmetic operators

| Operator | Name | Meaning |
|----------|------|-------------|
| a + b     | addition | sum of a and b|
| a - b     | subtraction | difference of a and b|
| a * b     | multiplication | product of a and b|
| a / b     | true division | quotient of a and b|
| a // b    | floor division | quotient of a and b, removing fractional part|
| a % b     | modulus | remainder after division of a by b|
| a ** b    | exponentiation | a raised to power of b|


````python
# working with integers
a = 13
b = 3

# Output: a + b
print('a + b =', a + b)
# Output: a - b
print('a - b =', a - b)
# Output: a * b
print('a * b =', a * b)
# Output: a / b
print('a / b =', a / b)
# Output: a // b
print('a // b =', a // b)
# Output: a % b
print('a % b =', a % b)
# Output: a ** b
print('a ** b =', a ** b)
````

````python
# example with real numbers
a = 1.2
b = 8.75
# Output: a + b
print('a + b =', a + b)
# Output: a - b
print('a - b =', a - b)
# Output: a * b
print('a * b =', a * b)
# Output: a / b
print('a / b =', a / b)
# Output: a // b
print('a // b =', a // b)
# Output: a % b
print('a % b =', a % b)
# Output: a ** b
print('a ** b =', a ** b)
````

## Comparison operators

| Operator | Meaning |
|----------|---------|
| >	| Greater than - True if left operand is greater than the right |
| <	|Less that - True if left operand is less than the right |
| == |	Equal to - True if both operands are equal |
| != |	Not equal to - True if operands are not equal |
| >= |	Greater than or equal to - True if left operand is greater than or equal to the right |
| <= |	Less than or equal to - True if left operand is less than or equal to the right |

````python
a = 13
b = 3.6

# Output: a > b
print('a > b :', a > b)
# Output: a < b
print('a < b :', a < b)
# Output: a == b
print('a == b :', a == b)
# Output: a != b
print('a != b :', a != b)
# Output: a >= b
print('a >= b :', a >= b)
# Output: a <= b
print('a <= b :', a <= b)
````

## Logical operators

Logical operators are the `and`, `or`, `not` operators.

| Operator | Meaning |
|----------|---------|
| and | True if both the operands are true |
| or | True if either of the operands is true |
| not | True if operand is false (complements the operand) |

````python
a = True
b = False

# Output: a and b
print('a and b :', a and b)
# Output: a or b
print('a or b :', a or b)
# Output: not a
print('not a :', not a)

# what happens with numbers?
a = 1.5
b = 0

# Output: a and b
print('a and b :', a and b)
# Output: a or b
print('a or b :', a or b)
# Output: not a
print('not a :', not a)
````

## Special operator

Identity operator `is` is used to check if two values (or variables) are located on the same part of the memory. Two variables that are equal does not imply that they are identical.

| Operator | Meaning |
|----------|---------|
|is | True if the operands are identical (refer to the same object) |


````python
a = 9
b = a
print('a is b', a is b)
print(id(a))
print(id(b))
````

````python
a = 9
b = 9
print('a == b', a == b)
````

````python
a = 9.9
b = 9.9
print('a is b', a is b)
print('id(a)', id(a))
print('id(b)', id(b))
print('a == b', a == b)
````

## Bit operators

| Operator | Meaning | Description |
|----------|---------|-------------|
|`a & b`  | Bitwise AND | Bits defined in both a and b |
|`a \| b`  | Bitwise OR | Bits defined in a or b or both |
|`a ^ b`  | Bitwise XOR | Bits defined in a or b but not both|
|`a << b`  | Shift left | Shift bits of a left by b units |
|`a >> b`  | Shift right | Bits defined in both a and b |
|`~a`  | Bitwise NOT | Bitwise negation of a|


````python
# example taken from https://www.tutorialspoint.com/python/bitwise_operators_example.htm

a = 60            # 60 = 0011 1100 
b = 13            # 13 = 0000 1101 
print('a = ', a, ' bin(a) = ', bin(a))
print('b = ', b, ' bin(b) = ', bin(b))
````

````python
c = a & b;        # 12 = 0000 1100
print ('Value of c is ', c)
````

````python
c = a | b;        # 61 = 0011 1101 
print ('Value of c is ', c)
````

````python
c = a ^ b;        # 49 = 0011 0001
print ('Value of c is ', c)
````

````python
c = a << 2;       # 240 = 1111 0000 - a multiplied by 4
print ('Value of c is ', c)
````

````python
c = a >> 2;       # 15 = 0000 1111 - a divided by 4
print ('Value of c is ', c)
````

````python
c = ~a;           # -61 = 1100 0011
print ('Value of c is ', c)
````
