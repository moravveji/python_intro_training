## integer
- Most basic numerical type. 
- Any number without a decimal point is an integer.
- Note: Python integers are variable-precision, not limited as in C, Matlab to 4 or 8 bytes.
    - see https://rushter.com/blog/python-integer-implementation/

```python
a = 2
b = 2**200
print('type of a', type(a))
print('type of b', type(b))
print('a: ', a)
print('b: ', b)
print('size of a:', a.__sizeof__(), ' bits')
print('size of b:', b.__sizeof__(), ' bits')
```

## floating point 

- The floating-point type can store fractional numbers. 
- standard decimal notation, or in exponential notation
  * ´a = 0.000005´
  * ´b = 5e-6´
 - Note:  limited precision, never rely on exact equality tests with floating-point values.
0.1 + 0.2 == 0.3


```python
a = 0.000005
b = 5.08e-6
print(a, b)
print('type', type(a))
print('type', type(b))
```

comparisons with float numbers can be tricky

due to the limited precision of representing floats, never rely on exact equality tests with floating-point values. 0.1 + 0.2 == 0.3

```python
a = 0.1
b = 0.2
print('a + b =', a + b)
c = 0.3
print('test a + b == c: ', (a + b) == c)
```

A better way to cmpare floats: check the absolute difference


```python
epsilon = 0.00001
x = 0.1 + 0.2
y = 0.3
print(abs(x - y) < epsilon)  
```

**Optional:** Get an idea of the binary representation of a float: you can convert the float to an integer using the int.from_bytes() function, and then convert that integer to binary with the bin() function.
    

```python
import struct

x = 0.1
print(x.hex())
packed = struct.pack('f', x)
integer = int.from_bytes(packed, byteorder='big')
binary = bin(integer)
print(binary)
y = x

# precision needed beyond 16th digit!
x = 0.10000000000000001
print(x.hex())
packed = struct.pack('f', x)
integer = int.from_bytes(packed, byteorder='big')
binary = bin(integer)
print(binary)

print(x==y)
```

In the context of programming and numerical computation, “machine epsilon” refers to the smallest representable number such that 1.0 + epsilon is not equal to 1.0. It represents the upper bound on the relative error due to rounding in floating point arithmetic.

Machine epsilon is a measure of precision in the representation of real numbers in a computer. It varies between systems and is specific to the floating-point format used by the system.

```python
import sys
epsilon = sys.float_info.epsilon
print(epsilon)
```

### strings and Methods 

`rstrip`

- Getting rid of line endings: `rstrip('\r\n')`
- Method will strip all combinations of \r and \n from right end of string
- Similar methods:
   - `lstrip`: strips from left end of string
   - `strip`: strips from both ends of string
- No arguments, strips: space, \t, \r, \n, …
- These methods do not modify the original string, instead they return a new string that has been altered


```python
str1 = 'this is a string'
print(dir(str1))
```

```python
str1.capitalize()
```

```python
my_str_1 = 'this is my string,\t with some text in it \n More text on a next line \n'
dir(my_str_1)
print(my_str_1)
print(my_str_1.rstrip())
```

`split`

Splitting string: `split()` returns list of strings
- no argument: split on (multiple) whitespace
- otherwise, split on provided string
- limit number of splits by providing extra argument


```python
my_str_1 = 'this is my string,\t with some text in it \n More text on a next line \n'
print(my_str_1)
my_str_split_1 = my_str_1.split()
print(my_str_split_1)
my_str_split_2 = my_str_1.split(',')
print(my_str_split_2)

my_str_3 = 'end 1: 2013-03-28 03:05:57'
# split on :, but keep the time format!
my_str_split_3 = my_str_3.split(':', 1)
print(my_str_split_3)
```

```python
my_long_string = ' this is a nice piece of text for you   '
my_words = my_long_string.split()
print(my_words)

# split on a specified character
my_words_bis = my_long_string.split('i')
print(my_words_bis)

#%% putting things together
new_word = ' '.join(my_words_bis).strip()
print(new_word)
```

### searching and replacing

- Function `len()` computes str length 
- replace with `replace`
    
Make combinations
- cut to pull out parts

```python
# substring in string
print('ab' in 'ABCD')
print('ab' in 'ABCDab')
```

```python
# extract information between braces
str1 = ' this is a string containing (some) text'
# find the first occurence of '('
start_brc = str1.index('(')
# cut from the first brace
tail = str1[start_brc+1:]
# find the first occurence of ')'
end_brc = tail.index(')')
text = tail[:end_brc]
print('extracted text: ', text)
print('len(str1) :', len(str1))
print(str1.replace('is', 'as'))
str1 = str1.replace('is', 'was')
print(str1)
```

## format output

### printf-style formatting: 
This is similar to the formatting used in C’s printf function. It’s less commonly used in modern Python code.

### str.format(): 
This method is more versatile and can be used in Python 2.7 and above.

### string Interpolation (f-strings): 
Available since Python 3.6
F-string is a string literal that is prefixed with `f` or `F`. These strings may contain replacement fields (delimited by curly braces {} – fill out the braces). F-string is evaluated at run time.

File: fstring_01.py


```python
name = 'Peter'
age = 23

print('%s is %d years old' % (name, age))       # printf style
print('{} is {} years old'.format(name, age))   # .format
print(f'{name} is {age} years old')             # f-string
```

```python
fnum = 2007.12345
print(f"Simple = {fnum}")
print(f"Decimal Places specified = {fnum:.2f}")
print(f"Significant Figures={fnum:.3g}")
```

## docstring

Documentation strings or “docstrings” use a special form of comment.
The lines are enclosed in triple double quotes 
"""  """
- Everything within triple double quotes is treated as a literal string and a comment, including line breaks.
- Docstrings are placed at the top of program units, just under the declaration of the unit name (if present). 
- If they are correctly placed, certain automated tools are available to display the documentation.
