|------|---------|---------|
| list | `[1, 2, 3]` | Ordered collection|

| dict | `{'a':1, 'b':2, 'c':3}` | Unordered (key:value) pair mapping|

# Tuples
Tuples are in many ways similar to lists, but they are defined with parentheses rather than square brackets:
- ()
- Tuples are defined by specifying items separated by commas
- immutable objects, they are usually used when the list of values doesn't change
- same slicing rules apply as for lists

Key features:
  - **heterogeneous**: elements of a tuple don't have to be the same type. Each element of the tuple can be of any type.
  - **indexing**: elements can be referenced by an index. 
  - **nesting**: tuples can be nested to arbitrary depth
  - **immutable**: tuples are unchangeable
  - **duplicates allowed**: tuples are indexed, items can have the same value

Tuples?
- if data do not change, implementing it as tuple will guarantee that it remains write-protected.
- tuples are immutable, iterating through tuple is faster than with list. So there is a slight performance boost.
- tuples are frequently used to return multiple variables from functions.


```python
# tuple of strings
my_tuple1 = ('hi', 'hello', 'world')
print(my_tuple1)

# tuple mixed data
my_tuple2 = (1, 'Hello', 3.14)
print(my_tuple2)

# tuple of string and list
my_tuple3 = ('moving', [1, 2, 3, 4], ('v', 't', 'y'))
print(my_tuple3)
```

Watch out: mutable elements from a tuple can be changed

```python
#mutable elements from a tuple can be changed
my_tuple3[1][0] = 100
print('my_tuple3', my_tuple3)

# tuple with 1 element
my_tuple4 = (1,)
print(my_tuple4)

# empty tuple
my_tuple5 = ()
print(my_tuple5)
```

select items from a tuple with []

```python
#select elements out of the tuples
print(my_tuple1[1])
print(my_tuple1[1][1])
print(my_tuple2[1])
print(my_tuple3[2])
print(my_tuple3[2][2])
```

### packing / unpacking

Tuples can be used to assign multiple variables in a single operation (packing)

Syntax:  < tuple of variables > = < tuple of values >

```python
(a, b, c) = (1, 2, 3)
print(a, b, c)
print(type(c))

(a, b, *c) = (1, 2, 3, 4, 5)
print(a, b, c)
print(type(c))

(a, *b, c) = (1, 2, 3, 4, 5)
print(a, b, c)
print(type(c))

(a, b, c) = (1, 2, 3, 4, 5)
print(a, b, c)
print(type(c))
```

```python
t1 = (1, 2, 'help', [1, 2, 3])
print(t1)
print(dir(t1))
```

Methods

- index() searches an element in a tuple and returns its index.
- count() returns the number of items

```python
t1 = (1, 2, 'help', [1, 2, 3])
print(t1)
print('count of 1: ', t1.count(1))
print('count of 3: ', t1.count(3))

print('index of 1: ', t1.index(1))
print('index of 3: ', t1.index(3))
```

Test membership with `in`

```python
3 in t1
```

#### advanced

named tuple 

* Convenient access the tuple's fields by name, rather than by index. 
* Python's standard library: define named tuples through typing.NameTuple.

```python
from typing import NamedTuple

class Car(NamedTuple):
    name: str
    year: int
    weight: float
        

car1 = Car('Volvo', 2015, 48.9)
print('car = ', car1)
print('car name = ', car1.name)
```

```python
lcars = [Car('Volvo', 2015, 48.9), Car('VW', 2017, 4.9), Car('Opel', 2011, 58.9)]
print('car 1', lcars[0])
print('car 2', lcars[1].name)
```
