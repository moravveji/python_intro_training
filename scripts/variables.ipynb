### Mutable vs Immutable

* Immutability: Some of Python’s data types are immutable, which means they can’t be changed after they’re created. This includes `integers`, `floats`, `strings`, and `tuples`. If you try to change the value of an immutable variable, Python actually creates a new object with the new value.
* Mutable Types: Other data types in Python are mutable, which means they can be changed after they’re created. This includes `lists`, `dictionaries`, and `sets`.

```python
# Immutable
num = 5
print(num)
print(id(num))  # Let's say this prints: 140703167764848
num += 1
print(num)
print(id(num))  # This will print a different id: 140703167764880

# Mutable
lst = [1, 2, 3]
print(lst)
print(id(lst))  # Let's say this prints: 1856906640576
lst.append(4)
print(lst)
print(id(lst))  # This will print the same id: 1856906640576
```

### Global and Local Variables

* Variables defined inside a function are local to that function and cannot be accessed outside of it. 
* Variables defined outside all functions are global and can be accessed anywhere in the program.

* Local Variables:
  * A local variable is declared inside a function.
  * It is only accessible within the function where it is declared.
  * Different functions can use the same name for different local variables.
  * They are created when their function is called and destroyed when the function ends.

* Global Variables:
  * A global variable is declared outside all functions.
  * It is accessible throughout the entire program, including within functions (unless a local variable with the same name exists).
  * There can only be one global variable with a given name in a Python program.
  * They are created when their line of code is executed and exist until the program ends.

```python
# This is a global variable
x = 10

def my_function():
    # This is a local variable
    y = 5
    print("Inside the function, x = ", x, "and y = ", y)

my_function()

# This will print: Inside the function, x = 10 and y = 5
print("Outside the function, x = ", x)

# This will raise a NameError because y is not defined outside the function
print("Outside the function, y = ", y)
```
