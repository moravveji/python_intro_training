# Dictionaries
- {'a':1, 'b':2, 'c':3}
- **(key,value) mapping**: the key must be immutable, mostly a string. Dictionaries in Python are used to store data values in key-value pairs. Each key-value pair maps the key to its associated value.
- **mutable**: meaning that we can change, add, or remove items after the dictionary has been created.
- **nested dictionaries**: A dictionary can contain another dictionary. 
- **ordered?**: as of Python version 3.7, dictionaries are ordered. This means that the items have a defined order, and that order will not change. Note that in Python 3.5 and older dictionaries have no sense of order. In Python 3.6, dictionaries were stored in insertion order as an implementation improvement. In Python 3.7 and beyond, dictionaries are guaranteed to be ordered in the order in which their entries were created. It is therefore advisable be cautious when relying on ordering in dictionaries. For safety’s sake, it is better to assume there is no sense of order.
- **no indexing**: an index value as in `lists`and `tuples` cannot be used.
- **no duplicates**: dictionaries do not allow duplicate keys. If you try to use a duplicate key, the old value for that key will be replaced with the new value.
- **various data types**: the values in dictionary items can be of any data type.

The order in dictionaries is kept. 

In the example below, elements are added to the dictionary in a specific order: ‘first’, ‘second’, ‘third’, and ‘fourth’. When  the dictionary is printed, the elements are displayed in the order they were added. This is because, as of Python 3.7, dictionaries maintain the insertion order of their elements.

```python
# Creating a dictionary
my_dict = {}

# Adding elements to the dictionary
my_dict['first'] = 1
my_dict['second'] = 2
my_dict['third'] = 3
my_dict['fourth'] = 4

# Printing the dictionary
for key, value in my_dict.items():
    print(f"{key}: {value}")
```

Methods

- clear(): clears a dictionary
- keys(): returns a list of keys
- values(): returns a  list of values
- pop(): removes a key, and returns its value


```python
zipdict = {'Hasselt': 3500,
 'Luik': 4000,
 'Heverlee': 3001,
 'Leuven': 3000,
 'Brussel': 1000,
 'Gent': 9000,
 'Antwerpen': 2000,
 'Brugge': 8000}

#Use the function keys() - 
keys = zipdict.keys()
print(type(keys))
print(keys)

values = zipdict.values()
print(values)

zipdict.pop('Gent')
print(zipdict)

zipdictbis = zipdict
print('zipdictbis: ', zipdictbis)
zipdictbis.clear()
print('zipdictbis: ', zipdictbis)

# alias?
print('zipdict: ', zipdict)
```

```python
basketdict = dict([ 
    ('Leuven', 'Bears'),
    ('Hasselt','Limburg United'),
    ('Aalst','Okapi'),
    ('Oostende','BC Oostende')])

#Use the function keys() - 
for city in basketdict.keys():
    if city[0] == 'L':
        print('city starting with L:', city)

# create a list out of the keys
keys = basketdict.keys()
alist = list(keys)
alist.sort()
# print a sorted list
print(alist)
```

```python
# create a histogram
# https://swcarpentry.github.io/python-second-language/06-dict/
numbers = [1, 0, 1, 2, 0, 0, 1, 2, 1, 3, 1, 0, 2]
count = {}
print(type(count))
for n in numbers:
    if n not in count:
        count[n] = 1
    else:
        count[n] = count[n] + 1
print(count)
print(count[1]) # 1 is a key, not an index
```

It is possible to create a nested dictionary: a dictionary within a dictionary. 

In this example, a dictionary is created for an observation, the observations are combined in a parent-dictionary.

```python
Obs01 = {'name': 'experiment 1', 'val': 23.33, 'temp': 100.0, 'press': 200.0, 'date':'2022-02-02'}
Obs02 = {'name': 'experiment 2', 'val': 27.00, 'temp': 110.0, 'press': 200.0, 'date':'2022-02-02'}
Obs03 = {'name': 'experiment 3', 'val': 27.88, 'temp': 120.0, 'press': 200.0, 'date':'2022-02-03'}
series_01 = {'t1': Obs01,
            't2': Obs02,
            't3': Obs03} # insert individual dictionaries in a dictionary
# print result of the 3rd observation
print('val = {}, temp = {}'.format(series_01['t3']['val'], series_01['t3']['temp']))
```

### aliasing


Because dictionaries are mutable, you need to be aware of aliasing (as with lists). Whenever two variables refer to the same dictionary object, changes to one affect the other. 


Use the dictionary copy method

```python
dict1 = {"key1": "value1", "key2": "value2"}
dict2 = dict1
dict2["key2"] = "WHY?!"
dict1
```

```python
import copy

dict1 = {"key1": "value1", "key2": "value2"}
dict2 = copy.deepcopy(dict1)
dict2["key2"] = "WHY?!"
dict1
```

### A dictionary is an implementation of a hash table

http://justinbois.github.io/bootcamp/2023/lessons/l09_dictionaries.html


Each entry in the dictionary is stored at a different location in memory. The dictionary itself also has its own address. 

Dictionaries use a hash function to do this. A hash function converts its input to an integer. 

Python’s built-in hash function can be used to convert the keys to integers.
Under the hood, Python then converts these integers to integers that could correspond to locations in memory. A collection of elements that can be indexed this way is called a hash table. This is a very common data structure in computing.

```python
# Create dictionary
my_dict = {'a': 6, 'b': 7, 'c':12.6}

# Find where they are stored
print(id(my_dict))
print(id(my_dict['a']))
print(id(my_dict['b']))
print(id(my_dict['c']))

# python hash function
print(hash('a'), hash('b'), hash('c'))
```
