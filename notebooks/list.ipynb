## Slicing lists, to create sublists

- Assigning to slices
- All slice operations return a new list containing the requested elements
- Syntax: < var name >[< index1 >:< index2 >[:step]]
    - index1: Optional. Starting index of the slice. Defaults to 0.
    - index2: Optional. The last index of the slice or the number of items to get. Defaults to len(sequence).
    - step: Optional. Extended slice syntax. Step value of the slice. Defaults to 1.


```python
l = list(range(1, 6))
print('l', l)
l_sub = l[2:4]
print('l_sub', l_sub)
l_sub = l[:4]
print('l_sub', l_sub)
l_sub = l[2:]
print('l_sub', l_sub)
l_sub = l[0:4:3]
print('l_sub', l_sub)
l_sub = l[::2]
print('l_sub', l_sub)
l_sub = l[4:1:-1]
print('l_sub', l_sub)

l_r = l[::-1]
print('l_r', l_r)

#create a copy
l_bis = l[:]
print('l_bis', l_bis)
```

Slice assignment is allowed

```python
l = list(range(1, 6))
print('l:', l)

# replace a chunk with the same number of items
print(l[::2])
l[::2] = ['a', 'b', 'c']  
print('l:', l)

# replace a chunk with a larger number of items
print(l[:3])
l[:3] = [999, 999, 999, 999, 999]
print('l:', l)

# replace a chunk with a smaller number of items
print(l[:3])
l[:3] = [55555]
print('l:', l)
```

## Removing elements from a list

* `remove`: delete a specific object in the list. Remove is the only one that searches object (not index based).
* `del`: delete the object at a specific location (index) in the list.
* `pop`: delete the object at a specific location (index) in the list and get the object at the specific location.

```python
# remove elements by their position in a list, 
# or by the value of the item.  Python removes only the first occurrence of that value"
zips = [3050, 3001, 8500, 4000, 1000, 2000, 8000, 3500]

print('zips: ', zips)
del zips[1]
print('zips: ', zips)
del zips[1:4]
print('zips: ', zips)
zips.remove(8000)
print('zips: ', zips)
zips.remove(8000)
```

##  aliasing

If variable v1 refers to an object and an assignment is made `v2 = v1`, then both variables refer to the same object.


It’s important to note that any modifications to the alias object will reflect in the existing object and vice versa, because they both point to the same memory location.

If the aliased object is mutable, changes made with one alias affect the other!

```python
v1 = [1, 2, 3]
v2 = v1
print('v1 = ', v1)
print('v2 = ', v2)
print('v1 is v2 ? ', v1 is v2)
print('v1 == v2 ? ', v1 == v2)
print("Reference of 1st list:", id(v1))
print("Reference of 2nd list:", id(v2))

v2[1] = 101
print('v1 = ', v1)
print('v2 = ', v2)
```

### shallow copy

A partial solution is to use the list constructor or the slice operation to make a new list via a shallow copy. But if the list contains other compound types then they won't be independently copied


```python
# use slicing notation to create a shallow copy
v1 = [1, 2, 3]
v2 = v1[:]
print('v1 = ', v1)
print('v2 = ', v2)
print('v1 is v2 ?', v1 is v2)
print("Reference of 1st list:", id(v1))
print("Reference of 2nd list:", id(v2))

v1[1] = 101
print('v1 = ', v1)
print('v2 = ', v2)
```

```python
# use the list function to create a shallow copy
v1 = [1, 2, 3]
v2 = list(v1)
print('v1 = ', v1)
print('v2 = ', v2)
print('v1 is v2 ?', v1 is v2)
print("Reference of 1st list:", id(v1))
print("Reference of 2nd list:", id(v2))

v1[1] = 101
print('v1 = ', v1)
print('v2 = ', v2)
```

```python
# use the copy function to create a shallow copy
v1 = [1, 2, 3]
v2 = v1.copy()
print('v1 = ', v1)
print('v2 = ', v2)
print('v1 is v2 ?', v1 is v2)
print("Reference of 1st list:", id(v1))
print("Reference of 2nd list:", id(v2))

v1[1] = 101
print('v1 = ', v1)
print('v2 = ', v2)
```

```python
# be careful with a shallow copy on compound types!
# example taken from: https://cvw.cac.cornell.edu/python-intro/objects-and-oop/lists

myList = [[1,2,3],4]
print(myList)
myOtherList = myList.copy() # use the copy method of the list, is a shallow copy

myList[1] = 'four'
print(myList)
print (myOtherList)

#OK so far, but now try to change an element in the sublist
myList[0][1] = 'two'
print (myList)

#This time the item also changes in the shallow copy, because it was in a sublist
print (myOtherList)
```

Safer to avoid aliasing when working with mutable objects.

The way to make a completely independent copy of a list —a deep copy— is to use the the `copy.deepcopy()` method. 
`copy.deepcopy()` to makes copies of compound objects, as the errors which can occur from shallow copies are often frustrating to debug.

Python’s standard library provides the copy module, which provides functions that can be used to create deep copies of objects.

### list comprehension

Comprehension: powerful functionality within a single line of code; provides a compact way to create lists. 

The general syntax of list comprehension is: `[expression for item in list if condition]`. The if condition part is optional. List comprehension provides a more concise and readable way to create lists compared to using loops.

Works also for dictionaries and sets

```python
sqlist = [x * x for x in range(15)]
print(sqlist)
```

```python
# with if statement
even_numbers = [num for num in range(1, 10) if num % 2 == 0]
print(even_numbers)  # Output: [2, 4, 6, 8]
```

```python
# using if else
numbers = [1, 2, 3, 4, 5, 6]
even_odd_list = ["Even" if i % 2 == 0 else "Odd" for i in numbers]
print(even_odd_list)  # Output: ['Odd', 'Even', 'Odd', 'Even', 'Odd', 'Even']
```
