### if statement

If expression is True, execute all statements **indented** underneath
*note the indents and the colon :*
```python
if expression:
    statement(s)
```

````python
if 0==0:
    print('Hello World')
````

````python
if 0==1:
    print('You should never see me!')
````

````python
# change the value of x
x = 0

if (x > 2):
   print('Hello')
print('World')
````

### if else statement

If expression is True, execute all statements indented under if.
If expression is False, execute all statements indented under else.

```python
if expression:
    statement(s)
else:
    statement(s)
```

````python
x = 0

if (x > 2):
   print('Hello')
else:
    print('No')
    print('World')
````

## if elif else statement
Execute a code block only if some condition holds


**syntax**
    
*note the indents and the colon :*
```python
if expression:
    statement(s)
elif expression:
    statement(s)
elif expression:
    statement(s)
else:
    statement(s)
```

`elif` and `else` blocks are optional

- no limit on number of `elif`
- Booleans are checked in order. Once it finds first True, skips over all others


````python
x = -100

if x == 0:
    print(x, 'is zero')
elif x > 0:
    print(x, 'is positive')
elif x < 0:
    print(x, 'is negative')
else:
    print(x, 'what happend here?')
````

## nesting

With nesting, you combine the expressions.

Beware that with nested conditionals, your workflow/logic gets more and more complex, and may require substantial testing of each condition.

````python
temp = 60
pressure = 100

if 20 <= temp < 40:
    if pressure <= 60:
        print('Case A')
    else:
        print('Case B')
elif temp >80:
    print('Case C')

else:
    print('don\'t know')
````

## Exceptions

An exception is an event that arises when an error occurs during the execution of a program. Python uses exception objects to signal that something has gone wrong. So, consider exceptions as informative helpers, than scary messages! Normally, the exception name offers a hint on what has gone wrong and the exception traceback points at the exact line of code where things has gone wrong. However, when investigating/resolving issues in a given line, most probably you have to look around in the lines before that.

Python has an excellent collection of *built-in* exceptions, but some examples are:

- `Exception`: This is the base class for most error types.
- `ArithmeticError`: This is the base class for exceptions that occur for numeric calculations.
- `ZeroDivisionError`: This is a subclass of ArithmeticError and is raised when you’re trying to divide by zero.
- `FileNotFoundError`: This exception is raised when a file or directory is requested but doesn’t exist.


````python
# Doing it in a brute force way
z = 1 / 0
````

````python
# Doing it decently with try-except-finally to catch an exception (ZeroDivisionError in this case)
# E.g. somewhere in the flow of your program, you might encounter a division by zero situation
numerator = 1
denominator = 0

try:
    x = numerator / denominator
except ZeroDivisionError as e:
    print(f'Caught an exception: {e}')
finally:     # the 'finally' block always executes, regardless of whether an exception was caught or not and is optional
    print(f'The try-except clause is exhausted now')
````
