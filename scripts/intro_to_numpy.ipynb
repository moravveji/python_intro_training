## A very brief introduction to NumPy arrays

The central object for NumPy and SciPy is  `ndarray`, "NumPy array." This is an array object that is convenient for scientific computing. 

NumPy arrays are homogeneous, meaning all elements must be of the same data type. They are similar to Python lists but offer more functionality, especially for numerical operations.


## Create an array

There are several mechanisms for creating arrays as mentioned in the NumPy documentation:

* Conversion from other Python structures (i.e. lists and tuples)
* Intrinsic NumPy array creation functions (e.g. arange, ones, zeros, etc.)
* Replicating, joining, or mutating existing arrays
* Reading arrays from disk, either from standard or custom formats
* Creating arrays from raw bytes through the use of strings or buffers
* Use of special library functions (e.g., random)


### Conversion from other Python structures

convert to a NumPy array using `np.array()`

NumPy arrays can be defined using Python sequences such as lists and tuples. 

```python
# Create a 1D numpy array
arr1 = np.array([10, 20, 30, 40])

print(arr1)
print(type(arr1))
```

```python
# Create a 2D numpy array
arr2 = np.array([[10,20,30],[40,50,60]])

print(arr2)
print(type(arr2))
```

```python
# Create a 2D numpy array
arr3 = np.array([[[10, 20], [30, 40]], [[50, 60], [70, 80]]])

print(arr3)
print(type(arr3))
```

```python
# Create a 1D numpy array
arrt1 = np.array((10.2, 20, 30, 40))

print(arrt1)
print(type(arrt1))

# Create a 2D numpy array
arrt2 = np.array(((10, 20, 30, 40),(50.2,60,70,80)))

print(arrt2)
print(type(arrt2))

# Create a 2D numpy array
arrt3 = np.array((((10, 20), (30, 40.2)), ((50, 60), (70, 80))))

print(arrt3)
print(type(arrt3))
```

When using numpy.array to define a new array, consider the `dtype` of the elements in the array, which can be specified explicitly.

```python
arr1_int8 = np.array([120, 121, 122], dtype=np.int8)
print(arr1_int8)
print(type(arr1_int8))
```

### Reshaping, concatenating  arrays
Once arrays are been created, these can be transformed to create new arrays. 
When you assign an array or its elements to a new variable, you have to explicitly numpy.copy the array, otherwise the variable is a view into the original array.

```python
# reshape with shape 
vec1 = np.linspace(3, 9, 10)
print(vec1)

vec1.shape = (2,5)
print(vec1)

# a new 2D array using the reshape() method
arr_vec1 = vec1.reshape((5, 2))
print(arr_vec1)
```

Concatenating arrays

The `np.concatenate()` function accomplishes this. We simply have to pass it a tuple containing the NumPy arrays we want to concatenate.

```python
arr2 = np.array([[10,20,30],[40,50,60]])
arr2b = np.array([[101,201,301],[401,501,601]])

combi_1 = np.concatenate((arr2,arr2b))
print(combi_1)

combi_2 = np.concatenate((arr2,arr2b), axis=1)
print(combi_2)
```

## Indexing

Getting access to the 1D numpy array is similar to lists or tuples, it has an index to indicate the location. It is 0-based, and accepts negative indices for indexing from the end of the array.



```python
arr = np.array([1, 4, 3])

# get an element at a specific index
print(arr[1])

# linspace(a,b,n) generates an array of n equally spaced elements starting from a and ending at b.
vec1 = np.linspace(3, 9, 10)
print(vec1)

print(vec1[-1])
```

It is not necessary to separate each dimension’s index into its own set of square brackets.

```python
arr2 = np.array([[10,20,30],[40,50,60]])

print(arr2[1,2])
print(arr2[1][2])

# stay between the boundaries
print(arr2[2,3])
```

## NumPy arrays are mutable

NumPy arrays are mutable. 

```python
# Make an array
my_ar = np.array([1, 2, 3, 4])

# Change an element
my_ar[2] = 6

# See the result
my_ar
```

Now, let's try attaching another variable to the NumPy array.

```python
# Attach a new variable
my_ar2 = my_ar

# Set an entry using the new variable
my_ar2[3] = 9

# Does the original change? (yes.)
my_ar
```

Let's see how messing with NumPy in functions affects things.

### Create Array from Existing Array
Create an array from an existing array by copying array elements into the other array.

Using the `copy()` Method
To create an array from an existing NumPy array Python provides an in-built method that is the copy() method. In simpler words to copy the array elements into another array. If you make changes in an original array that will not be reflected in a copy method. 

```python
# Create array from existing array
# Using copy()
arr_0=np.array([10,20,30])
print("Original array",arr_0)
print('id arr_0', id(arr_0))

arr_1 = arr_0
print('arr_1', arr_1)
print('id arr_1', id(arr_1))

arr_2=arr_0.copy()
print('arr_2', arr_2)
print('id arr_2', id(arr_2))
```

### Slices of NumPy arrays are **views**, not copies

A very important distinction between NumPy arrays and lists is that slices of NumPy arrays are **views** into the original NumPy array, NOT copies.

```python
# Make list and array
my_list = [1, 2, 3, 4]
print(my_list)
my_ar = np.array(my_list)
print(my_ar)

# Slice out of each
my_list_slice = my_list[1:-1]
print(my_list_slice)
my_ar_slice = my_ar[1:-1]
print(my_ar_slice)

# Mess with the slices
my_list_slice[0] = 9
my_ar_slice[0] = 9

# Look at originals
print(my_list)
print(my_ar)
```

## Mathematical operations with arrays

Mathematical operations on arrays are done elementwise to all elements

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Element-wise addition
c = a + b
print(c)

# Element-wise multiplication
d = a * b
print(d)

# Element-wise division
e = a / b
print(e)
```

```python
# Multiply by scalar
-4 * a
```

```python
# Raise to power
b**2
```

matrix operations like dot products.

```python
help(np.dot)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[1, 1], [1, 1]])
c = np.dot(a, b)
print(c)


a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.dot(a, b)
print(c)
```
