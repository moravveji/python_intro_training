# for loops
repeat a code block

definite iteration: the number of repetitions is specified explicitly in advance

**(basic) syntax**

```python
for <var> in <iterable>:
    statement(s)
else:
    statement(s)
```
The indentation is very important!
The `else` clause gets entered only if the for-loop is fully exhausted.


**concept**

To carry out the iteration this for loop describes, Python does the following:
- Calls iter() to obtain an iterator
- Calls next() repeatedly to obtain each item from the iterator in turn
- Terminates the loop when next() raises the StopIteration exception


````python
sum=0
numbers =[1,22,31,45]
for num in numbers:
    print(num)
    sum=sum+num  # check the indentation
avg=sum/len(numbers)
print ('Average:', avg)
````

# while loops

Repeat a code block until some condition is met.

The argument of the while loop is evaluated as a boolean statement, and the loop is executed until the statement evaluates to False.
It is not always known beforehand how many times the loop will have to repeated.

**(basic) syntax**

```python
while condition:
    statement(s)
else:
    statement(s)
```
The indentation is very important! The `else` clause is executed as soon as the condition in the `while` loop evaluates to `False`. Beware of infinite loops!


````python
i = 0
while i < 10:
    print(i, end=' ')
    i += 1
else:
     print('\n the end \n')
print('this is the next statement')   
````

## fine-tuning loops
### Skipping loop iterations

`continue` : skips the remainder of the current loop and goes to the next iteration.

`break`: breaks-out of the loop entirely and go to the next statement following the loop

#### continue

the loop iterates over a list of numbers. If the current number (num) is even (i.e., num % 2 == 0), the continue statement is executed. This causes the loop to skip the print(num) statement for even numbers and immediately move on to the next iteration. As a result, only the odd numbers are printed. 

````python
num_list = [1, 2, 3, 44, 56, 77, 8, 18, 19, 77.7, 100, 101, 23]
for num in num_list:
    if num%2 == 0:
        continue
    print(num)
````

#### break

the loop iterates over a list of numbers. If the current number (num) is even (i.e., num % 2 == 0), the break statement is executed. This causes the loop to break out of the loop when an even number is encountered.

````python
num_list = [1, 2, 3, 44, 56, 77, 8, 18, 19, 77.7, 100, 101, 23]
for num in num_list:
    if num%2 == 0:
        break    
    print(num)
````

````python
n = 101
while n < 1000:
    if not(n%20):
        break   
    print(n, 'in while loop')    
    n += 1
else:
    print('else part of the while statement')
print('out of while loop')
````

#### `pass`

The `pass` statement in Python is a placeholder statement that is used when the syntax requires a statement, but you don’t want to execute any code. It’s often used as a placeholder for future code.

The example below uses `pass`as an alternative for the `continue` statement.

````python
num_list = [1, 2, 3, 44, 56, 77, 8, 18, 19, 77.7, 100, 101, 23]
for num in num_list:
    if num%2 == 0:
        pass
    else:
       print(num)
````

### `else` part

**(full) syntax `for` loop**

```python
for <var> in <iterable>:
    statement(s) 
else: # optional 
    statement(s)
```

**(full) syntax `while` loop**

```python
while condition :
    statement(s) 
else: # optional 
    statement(s)
```
The indentation is very important!

**else** part

use case: implement search loops, searching for an item meeting a particular condition, and need to perform additional processing or raise warning if no acceptable value is found

- Loop should not contain `break` statement.
- The statements in the else block will be executed after all iterations are completed (normal termination).



````python
numbers =[1, 2, 32, 45]
for i in numbers: 
    print(i) 
    if i > 5.5:
       break
else: # Not executed as there is a break 
    print('No break encountered') 
print('finished')
````

# iterators


An iterator is a special object that gives values in succession, it implements the `next` protocol and raises `StopIteration` when exhausted.
Iterators:

- range()

- zip()

## range

If you need to iterate over a sequence of numbers, the built-in function `range()` can be used, it
generates arithmetic progressions

for loop is often used with the `range` function (a commonly-used iterator in Python)
* `range(n)` - sequence 0,1,…,n−1
* `range(m, n)` - sequence m,m+1,…,n−1
* `range(m, n, s)` - sequence m,m+s, m+2s,… <= n-1

Note: limitation of `range()`: it works only with integers. (consider NumPy `arange`). See also : https://pynative.com/python-range-for-float-numbers/



````python
for i in range(10):
    print(i, end=' ')
print()
    
for i in range(5):
    print('Hello World')
print('')

for _ in range(5):  # throwaway variable, the loop variable isn't used.
    print('Hello World')
````

In many ways the object returned by `range()` behaves as if it is a list, but it is an object
which returns the successive items of the desired sequence when you iterate over it, but it doesn’t really
make the list, thus saving memory.

Such an object is an *iterable*, that is, suitable as a target for functions and constructs that expect
something from which they can obtain successive items until the supply is exhausted.

* regular for loop uses the items
* use `enumerate` when index is needed. The function gives you back two loop variables:
    - The count of the current iteration
    - The value of the item at the current iteration
* use `zip` when skimming multiple lists

````python
fruitslist = ['banana', 'apple',  'mango', 'grape', 'orange']
# use of enumerate - tuple: (index, value)
for line in enumerate(fruitslist, start=0):
    print(line)    
print(' ')

# use of enumerate - individual elements
for num, color in enumerate(fruitslist, start=1):
    print(num, color)    
print(' ')

for num, color in enumerate(fruitslist, start=1):
    print(color)    
print(' ')

````

### note: generator

A generator is a function that produces a sequence of results instead of a single value. As generators do not store the sequence that is generated, they are memory-efficient. 

Many built in Python functions return generators to minimize use of memory - e.g. `range, zip, open` - A loop can be used to evaluate them.

````python
zz = zip(range(10), 'abcdefghij')
print(zz)

for i, c in zz:
    print(i, c)
````

### note: comprehension

check http://justinbois.github.io/bootcamp/2023/lessons/l15_comprehensions.html#List-comprehensions

Syntax sugar similar to that of generator expressions can be used to create lists, sets and dictionaries. A powerful functionality within a single line of code; provides a compact way to create lists, dictionaries, sets!

for a list:
`newlist = [expression for item in iterable if condition == True]`


````python
# list comprehension

lc1 = [i for i in range(10)]
print(lc1)

lc2 = [i**2 for i in range(10) if i % 2 == 1]
print(lc2)
````

````python
# dictionary comprehension

names = ['ann', 'jules', 'charles']
ages = [55, 44, 45]

dc = {a1: b1 for a1, b1 in zip(names, ages)}
print(dc)
dc
````

````python
#set comprehension

l1 = list(range(1,11))
sc = {element**3 for element in l1}
print(sc)
````
